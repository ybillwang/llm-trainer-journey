```python
C=torch.arange(9).reshape(3,3)
ver_bcast = C-C.mean(dim=1,dtype=torch.float32,keepdim=True)
```

```python
#torch.allclose 浮点数专用的“相等” 0.1 + 0.2 == 0.3 在 float64 下是 False（0.1 和 0.2 在二进制里都存不精确，加出来的其实是 0.30000000000000004）
ver_loop=torch.empty(3,3,dtype=torch.float32)
for i in range(C.shape[0]):
    ver_loop[i] = C[i].float() - C[i].float().mean()

assert torch.allclose(ver_bcast, ver_loop)   # 机器判定，过 = 验收达标
print("验证通过：广播版与双循环版一致")
```

##Day2小结
- 整型tensor使用mean前需要转换成float，否则会报错，所以做好直接用float类型的tensor。
- C.mean(dim=1,dtype=torch.float32,keepdim=True)如果去掉keepdim=True，会造成mean的shape是(3,)在broadcast进行右对齐的之后，补1，变成(1,3)，最后的结果是C中每一行各列的元素分别减了1,4,7，而不是第1行减1，第二行减4，第3行减7。这种情况下不报错，是非常隐蔽的错误。
- broadcast规则：
    - dimension首先右对齐
    - 从右往左，对应dimension上，要么数值一样，要么一个为1，要么一个不存在
